```python
#comaparative
a = 10
b = 3

igual = a == b
diferente = a != b
mayorQue = a > b
menorQue = a < b
mayorOIgual = a >= b
menorOIgual = a <= b

print("Igual: ",igual)
print("\nDiferene: ",diferente)
print("\nMayor que: ",mayorQue)
print("\nMenor que: ",menorQue)
print("\nMayor o igual: ",mayorOIgual)
print("\nMenor o igual: ",menorOIgual)
```
