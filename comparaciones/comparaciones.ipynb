```python
#comaparative
a = 10
b = 3

igual = a == b
diferente = a != b
mayorQue = a > b
menorQue = a < b
mayorOIgual = a >= b
menorOIgual = a <= b

print("Igual: ",igual)
print("Diferene: ",diferente)
print("Mayor que: ",mayorQue)
print("Menor que: ",menorQue)
print("Mayor o igual: ",mayorOIgual)
print("Menor o igual: ",menorOIgual)
```

```python
#asgination
x=7
print("1: ",x)
x+=2
print("2: ",x)
x-=2
print("3: ",x)
x*=2
print("4: ",x)
x/=2
print("5: ",x)
x%=2
print("6: ",x)
x//=2
print("7: ",x)
x**=2
print("8: ",x)
#inicializar x a 7 de nuevo
x = 7
x&=2
print("9: ",x)
x>>=2
print("10: ",x)
x<<=2
print("11: ",x)

```

```python
#logicos

a= 10
b= 3

resultadoAnd = (a > 5) and (b < 5) #True
resultadoOr = (a > 5) or (b < 5)   # True
resultadoNot = not (a > 5)         #False
print("Resultado And: ",resultadoAnd)
print("Relsultado Or: ",resultadoOr)
print("Resultado Not: ",resultadoNot)

a= True
b= False

print("Logical AND: ", a and b)
print("Logical OR: ", a or b)
print("Logical NOT a: ", not a)

a = 5 #representacion binaria 0101
b = 3 #representacion binaria 0011

resultado = a ^ b #resultado binario: 0110, 6 en decimal

print("binario: ", resultado)
```

```python
#math lib
import math
num = 50
raizCuadrada = math.sqrt(num)
print(f"La raiz cuadrada es: ", "{:,.2f}".format(raizCuadrada))
print(f"La raiz cuadrada es: {raizCuadrada:.2f}")

base = 2
exponente = 3
potencia = math.pow(base, exponente)
print("El",base,"elevado a la", exponente, "es:", potencia)

anguloGrados = 45
print("angulo de:",anguloGrados)
anguloRadians = math.radians(anguloGrados)
print("angulo radianes:",anguloRadians)
seno = math.sin(anguloRadians)
print("seno:",seno)
coseno = math.cos(anguloRadians)
print("coseno:",coseno)
tan = math.tan(anguloRadians)
print("tan:",tan)

numeroDecimal = 3.7
redondeoArriba = math.ceil(numeroDecimal)
redondeoAbajo = math.floor(numeroDecimal)
print("Redondeo hacia arriba de", numeroDecimal, "es:",redondeoArriba)
print("Redondeo hacia abajo de", numeroDecimal, "es:",redondeoAbajo)

#logaritmo natural (base e)
print("logaritmo natural:",math.log(math.e))

#logaritmo base 10
print("logaritmo base 10:",math.log10(100))
```
